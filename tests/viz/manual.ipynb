```python
import polars as pl
import rapidstats as rs
import plotnine as p9
```

```python
df = pl.DataFrame({"x": [0.2, 0.201, 0.5, 0.3, 0.7], "y": [0.8, 0.801, 0.4, 0.41, 0.2]})
```

```python
def plot(df):
    return p9.ggplot(df, p9.aes(x="x", y="y")) + p9.geom_point()
```

```python
plot(df)
```

```python
plot(rs.viz.thin_points(df, x="x", y="y", min_distance=0.05))
```

```python
plot(
    rs.viz.thin_points(
        df,
        x="x",
        y="y",
        min_distance=20,
        transform=rs.viz.ScreenTransform(width=640, height=480),
    )
)
```
